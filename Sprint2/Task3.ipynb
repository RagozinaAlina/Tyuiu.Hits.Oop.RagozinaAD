### Вариант задания 19


<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Subscription в C#, который будет представлять подписки на
различные услуги. На основе этого класса разработать 2-3 производных класса,
демонстрирующих принципы наследования и полиморфизма. В каждом из классов
должны быть реализованы новые атрибуты и методы, а также переопределены
некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Subscription:
• Атрибуты: ID подписки (SubscriptionId), Название услуги (ServiceName),
Стоимость подписки (Cost).

• Методы:

CalculateMonthlyCost(): метод для расчета ежемесячной стоимости
подписки.

ExtendSubscription(): метод для продления подписки на
дополнительный период.

GetSubscriptionDetails(): метод для получения деталей подписки.

Требования к производным классам:
1. ПодпискаНаОнлайнСервис (OnlineServiceSubscription): Должна содержать
дополнительные атрибуты, такие как Количество доступных пользователей
(MaxUsers). Метод CalculateMonthlyCost() должен быть переопределен для
учета количества пользователей при расчете стоимости.
2. ПодпискаНаСтreamинг (StreamingSubscription): Должна содержать
дополнительные атрибуты, такие как Количество одновременных потоков
(MaxStreams). Метод ExtendSubscription() должен быть переопределен для
добавления специальных предложений для продления подписки.
3. ПодпискаНаВидео(VideoSubscription) (если требуется третий класс): Должна
содержать дополнительные атрибуты, такие как Качество видео
(VideoQuality). Метод GetSubscriptionDetails() должен быть переопределен
для отображения качества видео вместе с другими деталями подписки.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование

```
public interface Discount
{
    void ApplyDiscount(int percent);
}

public interface Notification
{
    void SendNotification();
}

public abstract class Subscription
{
    public int SubscriptionID { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost { get; set; }
    public int DurationMonths { get; set; }
    public string OwnerName { get; set; }
    public bool IsActive { get; set; }

    public Subscription(int id, string name, decimal cost, int duration, string ownerName)
    {
        SubscriptionID = id;
        ServiceName = name;
        Cost = cost;
        DurationMonths = duration;
        OwnerName = ownerName;
        IsActive = true;
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost / DurationMonths;
    }

    public virtual void ExtendSubscription(int months)
    {
        DurationMonths += months; 
        Console.WriteLine($"Подписка '{ServiceName}' продлена на {months} мес. Всего: {DurationMonths} мес.");
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionID}, Услуга: {ServiceName}, Базовая стоимость: {Cost} руб/мес.");
    }

    public void Cancel()
    {
        IsActive = false;
        Console.WriteLine("Подписка отключена.");
    }

    public void changeOwner(string newOwner)
    {
        OwnerName = newOwner;
    }

    public void DoubleTheCost()
    {
        Cost = Cost * 2;
        Console.WriteLine($"Цена выросла до {Cost} руб.");
    }
}

public class OnlineServiceSubscription : Subscription
{
    public string Platform { get; set; }
    public int MaxUsers { get; set; }
    public double ServiceGB { get; set; }

    public OnlineServiceSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxUsers, double serviceGB)
        : base(id, name, cost, duration, ownerName)
    {
        MaxUsers = maxUsers;
        ServiceGB = serviceGB; 
        Platform = "PC"; 
    }

    public void ChangePlatform(string newPlatform)
    {
        Platform = newPlatform;
        Console.WriteLine($"Платформа изменена на {Platform}");
    }
}

public class streamingSubscription : Subscription
{
    public int MaxStreams { get; set; }

    public streamingSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxStreams)
        : base(id, name, cost, duration, ownerName)
    {
        MaxStreams = maxStreams;
    }
}

public class VideoSubscription : streamingSubscription
{
    public string VideoQuality { get; set; }
    public string Language { get; set; }

    public VideoSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxStreams, string videoQuality, string language)
        : base(id, name, cost, duration, ownerName, maxStreams)
    {
        VideoQuality = videoQuality;
        Language = language; 
    }

    public void Set()
    {
        VideoQuality = "4K";
    }

    public void ChangeLanguage(string language)
    {
        Language = language; 
    }
}

public class PremiumSubscription : VideoSubscription, Discount, Notification
{
    public PremiumSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxStreams, string videoQuality, string language)
        : base(id, name, cost, duration, ownerName, maxStreams, videoQuality, language) { }

    public void ApplyDiscount(int percent)
    {
        Cost = Cost - (Cost * percent / 100);
        Console.WriteLine($"Применена скидка {percent}%. Новая цена: {Cost} руб.");
    }

    public void SendNotification()
    {
        Console.WriteLine($"Здравствуйте, {OwnerName}! Ваша подписка {ServiceName} успешно обновлена.");
    }
}

OnlineServiceSubscription onlineSer = new OnlineServiceSubscription(1, "Office 365", 1000, 2, "Иван", 3, 15.0);
onlineSer.GetSubscriptionDetails();
onlineSer.ChangePlatform("Office 365 Premium");
onlineSer.DoubleTheCost();
Console.WriteLine();

VideoSubscription videoSub = new VideoSubscription(3, "YouTube", 1200, 1, "Владимир", 1, "4K", "Русский");
videoSub.GetSubscriptionDetails();
videoSub.ChangeLanguage("English");
videoSub.Set();
Console.WriteLine();

PremiumSubscription prSub = new PremiumSubscription(4, "Netflix VIP", 3000, 12, "Петр", 5, "4K", "Русский");
prSub.GetSubscriptionDetails();
prSub.ApplyDiscount(20);
prSub.SendNotification();
```
