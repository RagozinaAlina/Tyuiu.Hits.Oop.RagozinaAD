<h2 style="color:DodgerBlue">Описание проекта:</h2>

----

Создать базовый класс Subscription в C#, который будет представлять подписки на различные услуги. На основе этого класса разработать 2-3 производных класса, демонстрирующих принципы наследования и полиморфизма. В каждом из классов должны быть реализованы новые атрибуты и методы, а также переопределены некоторые методы базового класса для демонстрации полиморфизма.

Требования к базовому классу Subscription: • Атрибуты: ID подписки (SubscriptionId), Название услуги (ServiceName), Стоимость подписки (Cost).

• Методы:

CalculateMonthlyCost(): метод для расчета ежемесячной стоимости подписки.

ExtendSubscription(): метод для продления подписки на дополнительный период.

GetSubscriptionDetails(): метод для получения деталей подписки.

Требования к производным классам:

1. ПодпискаНаОнлайнСервис (OnlineServiceSubscription): Должна содержать дополнительные атрибуты, такие как Количество доступных пользователей (MaxUsers). Метод CalculateMonthlyCost() должен быть переопределен для учета количества пользователей при расчете стоимости.
2. ПодпискаНаСтreamинг (StreamingSubscription): Должна содержать дополнительные атрибуты, такие как Количество одновременных потоков (MaxStreams). Метод ExtendSubscription() должен быть переопределен для добавления специальных предложений для продления подписки.
3. ПодпискаНаВидео(VideoSubscription) (если требуется третий класс): Должна содержать дополнительные атрибуты, такие как Качество видео (VideoQuality). Метод GetSubscriptionDetails() должен быть переопределен для отображения качества видео вместе с другими деталями подписки.

#### Дополнительное задание
Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) исользуйтие в проекте коллекции, делегаты, события.

```
using System;
using System.Collections.Generic;
using System.Linq;

public interface Discount
{
    void ApplyDiscount(int percent);
}

public interface Notification
{
    void SendNotification();
}

public interface Cancellable
{
    void Cancel();
}

public interface ISubscriptionRepository<T> where T : Subscription
{
    void Add(T subscription);
    void Remove (int id);
    void ShowAllDetails();
}

public abstract class Subscription: Cancellable
{
    public int SubscriptionID { get; set; }
    public string ServiceName { get; set; }
    public decimal Cost { get; set; }
    public int DurationMonths { get; set; }
    public string OwnerName { get; set; }
    public bool IsActive { get; set; }
    public DateTime CreationDate { get; set; }
    public string Currency { get; set; }
    public string PromoCode { get; set; }

    public event Action<string> OnStatusChanged;

    public Subscription(int id, string name, decimal cost, int duration, string ownerName)
    {
        SubscriptionID = id;
        ServiceName = name;
        Cost = cost;
        DurationMonths = duration;
        OwnerName = ownerName;
        IsActive = true;
        CreationDate = DateTime.Now;
        Currency = "руб";
        PromoCode = "NONE";
    }

    protected void Notify(string message)
    {
        OnStatusChanged?.Invoke($"ID {SubscriptionID} - {ServiceName}: {message} ");
    }

    public virtual decimal CalculateMonthlyCost()
    {
        return Cost / DurationMonths;
    }

    public virtual void ExtendSubscription(int months)
    { 
        DurationMonths += months;
        Console.WriteLine($"Подписка '{ServiceName}' продлена на {months} мес. Всего: {DurationMonths} мес.");
    }

    public virtual void GetSubscriptionDetails()
    {
        Console.WriteLine($"ID: {SubscriptionID}, Услуга: {ServiceName}, Базовая стоимость: {Cost} руб/мес.");
    }

    public void changeOwner(string newOwner)
    {
        OwnerName = newOwner;
    }

    public void DoubleTheCost()
    {
        Cost = Cost * 2;
        Console.WriteLine($"Цена выросла до {Cost} руб.");
    }

    public void ApplyPromo()
    {
        Console.WriteLine($"Промокод не указан. Скидка 0%.");
    }

    public void ApplyPromo(string code, decimal discountAmout)
    {
        PromoCode = code;
        Cost = Math.Max(0, Cost - discountAmout);
        Console.WriteLine($"Применен промокод '{code}'. Скидка: {discountAmout} {Currency}. Новая стоимость: {Cost}.");
    }

    void Cancellable.Cancel()
    {
        IsActive =false;
        Console.WriteLine($"Подписка {SubscriptionID} ({ServiceName}) полностью деактивирована.");
    }
}

public class OnlineServiceSubscription : Subscription
{
    public string Platform { get; set; }
    public int MaxUsers { get; set; }
    public double ServiceGB { get; set; }

    public OnlineServiceSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxUsers, double serviceGB)
        : base(id, name, cost, duration, ownerName)
    {
        MaxUsers = maxUsers;
        ServiceGB = serviceGB; 
        Platform = "PC"; 
    }

    public override decimal CalculateMonthlyCost()
    {
        return base.CalculateMonthlyCost() + (MaxUsers*100);
    }

    public void ChangePlatform(string newPlatform)
    {
        Platform = newPlatform;
        Console.WriteLine($"Платформа изменена на {Platform}");
    }
}

public class streamingSubscription : Subscription
{
    public int MaxStreams { get; set; }

    public streamingSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxStreams)
        : base(id, name, cost, duration, ownerName)
    {
        MaxStreams = maxStreams;
    }

    public override void ExtendSubscription(int month)
    {
        base.ExtendSubscription(month);
    }
}

public class VideoSubscription : streamingSubscription
{
    public string VideoQuality { get; set; }
    public string Language { get; set; }

    public VideoSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxStreams, string videoQuality, string language)
        : base(id, name, cost, duration, ownerName, maxStreams)
    {
        VideoQuality = videoQuality;
        Language = language; 
    }

    public override void GetSubscriptionDetails()
    {
        base.GetSubscriptionDetails();
    }

    public new void ChangeOwner(string newOwner)
    {
        OwnerName = "Premium-Владелец: " + newOwner;
        Console.WriteLine($"Владелец изменен на: {OwnerName}");
    }
}

public class PremiumSubscription : VideoSubscription, Discount, Notification
{
    public PremiumSubscription(int id, string name, decimal cost, int duration, string ownerName, int maxStreams, string videoQuality, string language)
        : base(id, name, cost, duration, ownerName, maxStreams, videoQuality, language) { }

    public void ApplyDiscount(int percent)
    {
        Cost = Cost - (Cost * percent / 100);
        Console.WriteLine($"Применена скидка {percent}%. Новая цена: {Cost} руб.");
    }

    public void SendNotification()
    {
        Console.WriteLine($"Здравствуйте, {OwnerName}! Ваша подписка {ServiceName} успешно обновлена.");
    }
}

public class SubscriptionManager<T> : ISubscriptionRepository<T> where T : Subscription
{
    private readonly List<T> _subscriptions = new List<T>();

    private void HandleSubscriptionNotification(string message)
    {
        Console.WriteLine($"[Уведомление] {message}");
    }

    public void Add(T subscription)
    {
        _subscriptions.Add(subscription);
        subscription.OnStatusChanged += HandleSubscriptionNotification;
        Console.WriteLine($"Зарегистрирована подписка: {subscription.ServiceName}");
    }

    public void Remove(int id)
    {
        var sub = _subscriptions.FirstOrDefault(c => c.SubscriptionID == id);
        if (sub != null)
        {
            sub.OnStatusChanged -= HandleSubscriptionNotification;
            _subscriptions.Remove(sub);
            Console.WriteLine($"Подписка с ID {id} удалена.");
        }
    }

    public void ShowAllDetails()
    {
        foreach (var sub in _subscriptions)
        {
            sub.GetSubscriptionDetails();
        }
    }
}

public class SubscriptionApp
{
    private readonly ISubscriptionRepository<Subscription> _repo;

    public SubscriptionApp(ISubscriptionRepository<Subscription> repo)
    {
        _repo = repo;
    }

    public void Demo(Subscription sub)
    {
        _repo.Add(sub);
        _repo.ShowAllDetails();
    }
}

ISubscriptionRepository<Subscription> myArchive = new SubscriptionManager<Subscription>();
SubscriptionApp app = new SubscriptionApp(myArchive);

VideoSubscription youtubeSub = new VideoSubscription(1, "YouTube Premium", 1500, 1, "Владимир", 2, "4K", "Русский");

app.Demo(youtubeSub);

youtubeSub.ExtendSubscription(3); 
youtubeSub.ApplyPromo("PROMO", 200);

Cancellable cancellableSub = youtubeSub;
cancellableSub.Cancel(); 

myArchive.Remove(1);
```
